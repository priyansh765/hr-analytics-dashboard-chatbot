```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv')
df.head()
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
for col in df.columns:
    if df[col].nunique() == 1:
        print(col, "-> only one unique value:", df[col].unique())
```

```python
df = df.drop(columns=['EmployeeCount', 'Over18', 'StandardHours'], errors='ignore')
df.shape
```

```python
df['EmployeeNumber'].nunique() == len(df)
```

```python
categorical_cols = df.select_dtypes(include='object').columns.tolist()
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

print("Categorical columns:", categorical_cols)
print("\nNumerical columns:", numerical_cols)
```

```python
df['AgeGroup'] = pd.cut(df['Age'], bins=[18,25,35,45,55,65], 
                          labels=['18-25','26-35','36-45','46-55','56-65'])
df[['Age','AgeGroup']].head()
```

```python
df['SalaryBand'] = pd.cut(df['MonthlyIncome'], bins=[0,3000,6000,10000,15000,20000],
                            labels=['Low','Medium','High','Very High','Top'])
df[['MonthlyIncome','SalaryBand']].head()
```

```python
df['TenureGroup'] = pd.cut(df['YearsAtCompany'], bins=[-1,2,5,10,20,40],
                             labels=['0-2 yrs','3-5 yrs','6-10 yrs','11-20 yrs','20+ yrs'])
df[['YearsAtCompany','TenureGroup']].head()
```

```python
df['TenureGroup'] = pd.cut(df['YearsAtCompany'], bins=[-1,2,5,10,20,40],
                             labels=['0-2 yrs','3-5 yrs','6-10 yrs','11-20 yrs','20+ yrs'])
df[['YearsAtCompany','TenureGroup']].head()
```

```python
df.to_csv('../data/processed/hr_cleaned.csv', index=False)
print("Saved successfully!")
```
